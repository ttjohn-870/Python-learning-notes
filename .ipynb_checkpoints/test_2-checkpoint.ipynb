#### 2.3.1.8鸭子类型

#不去判别它是列表还是字符串，只检验能不能被for循环遍历

```python
isiterable([1,2,3])
```

```python
def isiterable(xx):
    try:
        iter(xx)
        return True
    except TypeError:# 不可遍历
        return False
x = "abc"
# 如果它不是列表并且可以迭代，可以将它转化成列表输出
if not isinstance(x,list) and isiterable(x):
    x = list(x)
print(x)
```

```python
def add_and_maybe_multiply(a,b,c = None):
    result = a + b
    if c is not None:
        result = result*c
    return result
print(add_and_maybe_multiply(3,4,5))
```

### 电商订单超时自动关闭例子

```python
#timedelta:专门用来代表一段时间长度，用datetime和datetime 相减得到
from datetime import datetime,timedelta
# 1、模拟：用户下单的时间
order_time = datetime(2026,10,12,15,0,0)
# 2、设置超时阈值：15分钟
expire_gap = timedelta(minutes=15)
# 3、用于获取此刻系统时间
now_time = datetime.now()
if now_time -order_time >expire_gap:
    print("订单超时，关闭订单")
else:
    print("请尽快付款")
```

### 2.3.3控制流

#### 2.3.3.1 if、elif和else

```python
# 一个if可以接多个elif和一个else代码块，如果所有的elif条件均为False，则执行else语句
# input：用来让使用者输出数字
week = int(input("请输入星期数字（1-7）:"))
if week >=1 and week <=5:
    print("工作日")
elif week == 6 or week == 7:
    print("休息日")
else : 
    print("输入错误")
```

#### 2.3.3.2 for循环

```python
sequence = [1,2,None,4,None,8]
total = 0
for value in sequence:
    if value is None:
        break
    total += value
print(total)
```

```python
sequence = [1,2,None,4,None,8]
total = 0
for value in sequence:
    if value is None:
        continue
    total += value
    # total += value表示total =total +value
print(total)
```

#### 2.3.3.3 while循环

```python
x = 256
total = 0
while x >0:
    if total >500:
        break
    total += x
    x = x//2
    #//表示整数除法
print(x)
```
